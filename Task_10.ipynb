# TASK 10

**(QN)** Implement RNN for sentiment analysis on movie reviews

# STEPS

1. Import packages
2. Load the imdb dataset
3. Define the RNN Architecture
3. Define losss, optimizer
4. Evaluate the model
5. Make predictions


```python
#Import packages

from keras.datasets import imdb
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding, Flatten

#from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
#Load the imdb dataset

(train_X,train_Y),(test_X,test_Y)=imdb.load_data()
```

```python
#Preprocessing

train_X = pad_sequences(train_X, padding = 'post', maxlen = 50)
test_X = pad_sequences(test_X, padding = 'post', maxlen = 50)
```

```python
train_X.shape
```

```python
#Define the RNN Architecture

model = Sequential()

model.add(SimpleRNN(32, input_shape = (50,1), return_sequences = False))
model.add(Dense(1, activation = 'sigmoid'))
```

```python
model.summary()
```

```python
#Define loss , optimizer

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
#train the model
model.fit(train_X, train_Y, epochs = 10, validation_data = (test_X,test_Y))
```

```python
#Evaluate the model

test_loss, test_acc = model.evaluate(test_X, test_Y)
print('Test loss', test_loss)
print('Test accuracy', test_acc)
```

```python
def make_predictions(model, data, true_labels, num_predictions=5):
  predictions = model.predict(data)
  print(f"Displaying the first {num_predictions} predictions:\n")
  for i in range(num_predictions):
    predicted_sentiment = 'Positive' if predictions[i][0] > 0.5 else 'Negative'
    true_sentiment = 'Positive' if true_labels[i] == 1 else 'Negative'
    print(f"Review {i+1}: Predicted: {predicted_sentiment} (Score: {predictions[i][0]:.4f}), True: {true_sentiment}")

# Call the prediction function with the test data
make_predictions(model, test_X, test_Y)
```
